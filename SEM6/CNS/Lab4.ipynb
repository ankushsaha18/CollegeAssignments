```python
def generateMatrix(key):
    # Convert Upper casev& replace j with i
    key = key.upper().replace("J", "I")
    matrix = []
    # Remove Duplicate Letter
    used = set()
    
    for char in key:
        if char not in used:
            # Append Alphate Letter
            matrix.append(char)
            used.add(char)

    for char in "ABCDEFGHIKLMNOPQRSTUVWXYZ":  
        if char not in used:
            # Append the Remaining Character
            matrix.append(char)

    # The 25 character string is placed row wise in a 5X5 matrix
    return [matrix[i:i+5] for i in range(0, 25, 5)]

def textPairing(text):
    # Convert Upper case & replace j with i
    text = text.upper().replace("J", "I")

    i = 0
    result = ""
    # split into pair
    while i < len(text):
        a = text[i]
        b = text[i+1] if i+1 < len(text) else 'X'

        if a == b:
            #add X for reapeated letters
            result += a + 'X'
            i += 1
        else:
            result += a + b
            i += 2
            

    #add X for odd length letters
    if len(result) % 2 != 0:
        result += 'X'

    return result

# Find the Position of Letter
def position(matrix, char):
    for i in range(5):
        for j in range(5):
            if matrix[i][j] == char:
                return i, j


def encryptPlayfair(pt, key):
    matrix = generateMatrix(key)
    print("\nKey Matrix")
    for i in matrix:
        print(i)
    text = textPairing(pt)
    print("\nDigraphs--")
    for i in range(0, len(text)-1, 2):
        print(text[i], text[i+1])
    ct = ""

    for i in range(0, len(text)-1, 2):
        a, b = text[i], text[i+1]
        r1, c1 = position(matrix, a)
        r2, c2 = position(matrix, b)

        #Same Row : replace with its Right
        if r1 == r2:  
            ct += matrix[r1][(c1 + 1) % 5]
            ct += matrix[r2][(c2 + 1) % 5]

        #Same Column : replace with its Below
        elif c1 == c2:  
            ct += matrix[(r1 + 1) % 5][c1]
            ct += matrix[(r2 + 1) % 5][c2]

        #Rectangle : replace each letter in same row but column of the other letter
        else:  
            ct += matrix[r1][c2]
            ct += matrix[r2][c1]

    return ct


def decryptPlayfair(ct, key):
    matrix = generateMatrix(key)
    print("\nKey Matrix")
    for i in matrix:
        print(i)
    text = ct.upper()
    print("\nDigraphs--")
    for i in range(0, len(text)-1, 2):
        print(text[i], text[i+1])
    pt = ""

    for i in range(0, len(text)-1, 2):
        a, b = text[i], text[i+1]
        r1, c1 = position(matrix, a)
        r2, c2 = position(matrix, b)

        #Same Row : replace with its Right
        if r1 == r2: 
            pt += matrix[r1][(c1 - 1) % 5]
            pt += matrix[r2][(c2 - 1) % 5]

        #Same Column : replace with its Below
        elif c1 == c2: 
            pt += matrix[(r1 - 1) % 5][c1]
            pt += matrix[(r2 - 1) % 5][c2]

        #Rectangle : replace each letter in same row but column of the other letter
        else:  
            pt += matrix[r1][c2]
            pt += matrix[r2][c1]

    return pt

pt = input("Enter plain text: ")
key = input("Enter key: ")
print("\nCipher Text: ",encryptPlayfair(pt, key))

ct = input("\n\nEnter cipher text: ")
key = input("Enter key: ")
print("\nPlain Text: ",decryptPlayfair(ct, key))
```

```python
def encryptVigenere(pt, key):
    # Convert plaintext to uppercase
    pt = pt.upper()  
    # Convert key to uppercase 
    key = key.upper() 
    
    nk = ""  
    i = 0

    # Generate repeating key equal to length of plaintext
    for _ in range(len(pt)):
        nk += key[i]
        i = (i + 1) % len(key)  
    ct = "" 

    for j in range(len(pt)):
        # Convert plaintext and key letters to number (0–25)
        p = ord(pt[j]) - ord('A')   
        k = ord(nk[j]) - ord('A')   

        # Apply Vigenere formula: (P + K) mod 26
        letter = chr((p + k) % 26 + ord('A'))

        ct += letter  

    return ct


def decryptVigenere(ct, key):
    # Convert ciphertext to uppercase
    ct = ct.upper()   
    # Convert key to uppercase
    key = key.upper() 

    nk = ""  
    i = 0

    # Generate repeating key
    for _ in range(len(ct)):
        nk += key[i]
        i = (i + 1) % len(key)

    pt = ""  

    for j in range(len(ct)):
        # Convert ciphertext and key letters to number
        c = ord(ct[j]) - ord('A')   
        k = ord(nk[j]) - ord('A')   

        # Apply Vigenere formula: (C - K) mod 26
        letter = chr((c - k + 26) % 26 + ord('A'))

        pt += letter  

    return pt


pt = input("Enter plain text: ")
key = input("Enter key: ")
print("\nCipher Text:", encryptVigenere(pt, key))

ct = input("\n\nEnter cipher text: ")
key = input("Enter key: ")
print("\nPlain Text:", decryptVigenere(ct, key))
```

```python
import string

# Generate Playfair Matrix
def generate_matrix(key, mode="row"):
    key = key.upper().replace("J", "I")
    result = []
    for ch in key:
        if ch not in result and ch.isalpha():
            result.append(ch)

    for ch in string.ascii_uppercase:
        if ch == 'J':
            continue
        if ch not in result:
            result.append(ch)

    matrix = [['' for _ in range(5)] for _ in range(5)]
    index = 0

    # Row-wise filling
    if mode == "row":
        for i in range(5):
            for j in range(5):
                matrix[i][j] = result[index]
                index += 1

    # Column-wise filling
    elif mode == "column":
        for j in range(5):
            for i in range(5):
                matrix[i][j] = result[index]
                index += 1

    return matrix


# Display Matrix
def display_matrix(matrix):
    for row in matrix:
        print(" ".join(row))


# Find Position of Character
def find_position(matrix, ch):
    for i in range(5):
        for j in range(5):
            if matrix[i][j] == ch:
                return i, j


# Prepare Plaintext
def prepare_text(text):
    text = text.upper().replace("J", "I")
    text = ''.join([c for c in text if c.isalpha()])
    prepared = ""
    i = 0

    while i < len(text):
        a = text[i]
        if i + 1 < len(text):
            b = text[i + 1]
            if a == b:
                prepared += a + 'X'
                i += 1
            else:
                prepared += a + b
                i += 2
        else:
            prepared += a + 'X'
            i += 1

    return prepared


# Encryption Function
def encrypt(plain_text, matrix):
    plain_text = prepare_text(plain_text)
    cipher_text = ""
    for i in range(0, len(plain_text), 2):
        a = plain_text[i]
        b = plain_text[i + 1]
        row1, col1 = find_position(matrix, a)
        row2, col2 = find_position(matrix, b)

        # Same row
        if row1 == row2:
            cipher_text += matrix[row1][(col1 + 1) % 5]
            cipher_text += matrix[row2][(col2 + 1) % 5]

        # Same column
        elif col1 == col2:
            cipher_text += matrix[(row1 + 1) % 5][col1]
            cipher_text += matrix[(row2 + 1) % 5][col2]

        # Rectangle rule
        else:
            cipher_text += matrix[row1][col2]
            cipher_text += matrix[row2][col1]

    return cipher_text


# Decryption Function
def decrypt(cipher_text, matrix):
    plain_text = ""
    for i in range(0, len(cipher_text), 2):
        a = cipher_text[i]
        b = cipher_text[i + 1]
        row1, col1 = find_position(matrix, a)
        row2, col2 = find_position(matrix, b)

        # Same row
        if row1 == row2:
            plain_text += matrix[row1][(col1 - 1) % 5]
            plain_text += matrix[row2][(col2 - 1) % 5]

        # Same column
        elif col1 == col2:
            plain_text += matrix[(row1 - 1) % 5][col1]
            plain_text += matrix[(row2 - 1) % 5][col2]

        # Rectangle rule
        else:
            plain_text += matrix[row1][col2]
            plain_text += matrix[row2][col1]

    return plain_text


# Simple Brute Force Cryptanalysis
def brute_force_attack(cipher_text, possible_keys):
    print("\nPossible Plaintexts:\n")
    for key in possible_keys:
        matrix = generate_matrix(key)
        decrypted = decrypt(cipher_text, matrix)
        print("Key:", key)
        print("Plaintext:", decrypted)
        print("------------------------")


# Main Program
key = input("Enter Secret Key: ")
plaintext = input("Enter Plaintext: ")

print("\nROW-WISE MATRIX\n")
row_matrix = generate_matrix(key, "row")
display_matrix(row_matrix)

cipher1 = encrypt(plaintext, row_matrix)
print("\nEncrypted Text (Row-wise):", cipher1)

print("\nCOLUMN-WISE MATRIX\n")
col_matrix = generate_matrix(key, "column")
display_matrix(col_matrix)

cipher2 = encrypt(plaintext, col_matrix)
print("\nEncrypted Text (Column-wise):", cipher2)

key = input("Enter Secret Key: ")
ciphertext = input("Enter Ciphertext: ")

print("\nROW-WISE MATRIX\n")
row_matrix = generate_matrix(key, "row")
display_matrix(row_matrix)

plain1 = decrypt(ciphertext, row_matrix)
print("\nDecrypted Text (Row-wise):", plain1)

print("\nCOLUMN-WISE MATRIX\n")
col_matrix = generate_matrix(key, "column")
display_matrix(col_matrix)

plain2 = decrypt(ciphertext, col_matrix)
print("\nDecrypted Text (Column-wise):", plain2)

# Example brute force attack
possible_keys = ["SECURITY", "NETWORK", "CRYPTO","KEY","MONARCHY"]
print("\nBRUTE FORCE CRYPTANALYSIS")
brute_force_attack(ciphertext, possible_keys)
```
