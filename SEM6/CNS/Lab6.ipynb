```python
import math
import random 

def keyGeneration():
    p = int(input("Enter first large prime p:"))
    q = int(input("Enter the second large prime q:"))
    n = p * q
    phi_n = (p-1) * (q-1)
    e = int(input('Choose e (1 < e < phi_n and gcd(e, phi_n)=1): '))
    while math.gcd(e, phi_n) != 1:
        print("Invalid e! Try again.")
        e = int(input('Choose e again: '))
    k = 1
    while (k * phi_n + 1) % e != 0:
        k += 1
    d = (k * phi_n + 1) // e

    return e, d, n

def encrypt(M, e, n):
    C = pow(M, e, n)
    return C

def decrypt(C, d, n):
    M = pow(C, d, n)
    return M
    
e, d, n = keyGeneration()

print("\nPublic Key (e, n):", (e, n))
print("Private Key (d, n):", (d, n))

M = int(input('\nEnter message (numeric) to encrypt: '))

C = encrypt(M, e, n)
print("Encrypted Ciphertext:", C)

M_decrypted = decrypt(C, d, n)
print("Decrypted Message:", M_decrypted)

if M == M_decrypted:
    print("Encryption & Decryption Successful")
else:
    print("Error in computation")
```

```python
import random

def diffie_hellman():
    
    # Prime number and root
    q = int(input("Enter a prime number q: "))
    p = int(input("Enter a primitive root p: "))
    
    # Private keys
    a = random.randint(1, q-1)  
    b = random.randint(1, q-1)  
    
    # Public keys
    A = pow(p, a, q)  
    B = pow(p, b, q)  
    
    print(f"A Public Key: {A}")
    print(f"B Public Key: {B}")
    
    # Shared secret computation
    secret_A = pow(B, a, q)
    secret_B = pow(A, b, q)
    
    print(f"Shared Secret A: {secret_A}")
    print(f"Shared Secret B: {secret_B}")
    
    if secret_A == secret_B:
        print("Key Exchange Successful")
    else:
        print("Key Exchange Failed")

    return secret_A


shared_secret = diffie_hellman()
print("Final Shared Secret Key:", shared_secret)
```
