```python
import string

# Generate Playfair Matrix
def generate_matrix(key, mode="row"):
    key = key.upper().replace("J", "I")
    result = []
    for ch in key:
        if ch not in result and ch.isalpha():
            result.append(ch)

    for ch in string.ascii_uppercase:
        if ch == 'J':
            continue
        if ch not in result:
            result.append(ch)

    matrix = [['' for _ in range(5)] for _ in range(5)]
    index = 0

    # Row-wise filling
    if mode == "row":
        for i in range(5):
            for j in range(5):
                matrix[i][j] = result[index]
                index += 1

    # Column-wise filling
    elif mode == "column":
        for j in range(5):
            for i in range(5):
                matrix[i][j] = result[index]
                index += 1

    return matrix


# Display Matrix
def display_matrix(matrix):
    for row in matrix:
        print(" ".join(row))


# Find Position of Character
def find_position(matrix, ch):
    for i in range(5):
        for j in range(5):
            if matrix[i][j] == ch:
                return i, j


# Prepare Plaintext
def prepare_text(text):
    text = text.upper().replace("J", "I")
    text = ''.join([c for c in text if c.isalpha()])
    prepared = ""
    i = 0

    while i < len(text):
        a = text[i]
        if i + 1 < len(text):
            b = text[i + 1]
            if a == b:
                prepared += a + 'X'
                i += 1
            else:
                prepared += a + b
                i += 2
        else:
            prepared += a + 'X'
            i += 1

    return prepared


# Encryption Function
def encrypt(plain_text, matrix):
    plain_text = prepare_text(plain_text)
    cipher_text = ""
    for i in range(0, len(plain_text), 2):
        a = plain_text[i]
        b = plain_text[i + 1]
        row1, col1 = find_position(matrix, a)
        row2, col2 = find_position(matrix, b)

        # Same row
        if row1 == row2:
            cipher_text += matrix[row1][(col1 + 1) % 5]
            cipher_text += matrix[row2][(col2 + 1) % 5]

        # Same column
        elif col1 == col2:
            cipher_text += matrix[(row1 + 1) % 5][col1]
            cipher_text += matrix[(row2 + 1) % 5][col2]

        # Rectangle rule
        else:
            cipher_text += matrix[row1][col2]
            cipher_text += matrix[row2][col1]

    return cipher_text


# Decryption Function
def decrypt(cipher_text, matrix):
    plain_text = ""
    for i in range(0, len(cipher_text), 2):
        a = cipher_text[i]
        b = cipher_text[i + 1]
        row1, col1 = find_position(matrix, a)
        row2, col2 = find_position(matrix, b)

        # Same row
        if row1 == row2:
            plain_text += matrix[row1][(col1 - 1) % 5]
            plain_text += matrix[row2][(col2 - 1) % 5]

        # Same column
        elif col1 == col2:
            plain_text += matrix[(row1 - 1) % 5][col1]
            plain_text += matrix[(row2 - 1) % 5][col2]

        # Rectangle rule
        else:
            plain_text += matrix[row1][col2]
            plain_text += matrix[row2][col1]

    return plain_text


# Simple Brute Force Cryptanalysis
def brute_force_attack(cipher_text, possible_keys):
    print("\nPossible Plaintexts:\n")
    for key in possible_keys:
        matrix = generate_matrix(key)
        decrypted = decrypt(cipher_text, matrix)
        print("Key:", key)
        print("Plaintext:", decrypted)
        print("------------------------")


# Main Program
key = input("Enter Secret Key: ")
plaintext = input("Enter Plaintext: ")

print("\nROW-WISE MATRIX\n")
row_matrix = generate_matrix(key, "row")
display_matrix(row_matrix)

cipher1 = encrypt(plaintext, row_matrix)
print("\nEncrypted Text (Row-wise):", cipher1)

print("\nCOLUMN-WISE MATRIX\n")
col_matrix = generate_matrix(key, "column")
display_matrix(col_matrix)

cipher2 = encrypt(plaintext, col_matrix)
print("\nEncrypted Text (Column-wise):", cipher2)

key = input("Enter Secret Key: ")
ciphertext = input("Enter Ciphertext: ")

print("\nROW-WISE MATRIX\n")
row_matrix = generate_matrix(key, "row")
display_matrix(row_matrix)

plain1 = decrypt(ciphertext, row_matrix)
print("\nDecrypted Text (Row-wise):", plain1)

print("\nCOLUMN-WISE MATRIX\n")
col_matrix = generate_matrix(key, "column")
display_matrix(col_matrix)

plain2 = decrypt(ciphertext, col_matrix)
print("\nDecrypted Text (Column-wise):", plain2)

# Example brute force attack
possible_keys = ["SECURITY", "NETWORK", "CRYPTO","KEY","MONARCHY"]
print("\nBRUTE FORCE CRYPTANALYSIS")
brute_force_attack(ciphertext, possible_keys)
```
