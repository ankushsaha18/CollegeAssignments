## 2.2 Task 2: Implement a Simple Blackjack Game

```python
import random
from enum import Enum

class Suit(Enum):
    HEARTS = "Hearts"
    DIAMONDS = "Diamonds"
    CLUBS = "Clubs"
    SPADES = "Spades"

class Card:
    def __init__(self, suit: Suit, rank: str):
        self.suit = suit
        self.rank = rank

    def __repr__(self):
        return f"Card({self.suit.value}, {self.rank})"

class Deck:
    RANKS = ["Ace", "2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King"]
    def __init__(self):
        self.cards = [Card(suit, rank) for suit in Suit for rank in self.RANKS]
        self.shuffle()

    def shuffle(self):
        random.shuffle(self.cards)

    def draw(self):
        return self.cards.pop() if self.cards else None

class Hand:
    """Manages a player's or dealer's hand and calculates its value."""
    def __init__(self):
        self.cards = []

    def add_card(self, card):
        self.cards.append(card)

    def get_value(self):
        """Calculates value based on Blackjack rules."""
        value = 0
        aces = 0
        for card in self.cards:
            if card.rank in ["Jack", "Queen", "King"]:
                value += 10 # Face cards are worth 10 points 
            elif card.rank == "Ace":
                aces += 1
                value += 11 # Initially count Ace as 11 
            else:
                value += int(card.rank)
        
        # Adjust for Aces if total exceeds 21
        while value > 21 and aces > 0:
            value -= 10
            aces -= 1
        return value

def play_blackjack():
    deck = Deck()
    player_hand = Hand()
    dealer_hand = Hand()

    # Deal initial two cards to both
    for _ in range(2):
        player_hand.add_card(deck.draw())
        dealer_hand.add_card(deck.draw())

    print(f"Player's Hand: {player_hand.cards} (Total: {player_hand.get_value()})")
    
    # Player's Turn
    while player_hand.get_value() < 21:
        action = input("Do you want to (h)it or (s)tand? ").lower()
        if action == 'h':
            player_hand.add_card(deck.draw())
            print(f"Player draws. Hand: {player_hand.cards} (Total: {player_hand.get_value()})")
        elif action == 's':
            break
        else:
            print("Invalid input. Please enter 'h' or 's'.")

    # Check if player busts
    if player_hand.get_value() > 21:
        print("Player goes bust! Dealer wins.")
        return

    # Dealer's Turn: must hit on 16 or below, stand on 17 or above 
    print(f"\nDealer's Hand starts as: {dealer_hand.cards} (Total: {dealer_hand.get_value()})")
    while dealer_hand.get_value() <= 16:
        print("Dealer hits...")
        dealer_hand.add_card(deck.draw())
        print(f"Dealer's Hand: {dealer_hand.cards} (Total: {dealer_hand.get_value()})")

    player_total = player_hand.get_value()
    dealer_total = dealer_hand.get_value()
    
    print("\n--- Final Results ---")
    if dealer_total > 21:
        print("Dealer busts! Player wins.")
    elif player_total > dealer_total:
        print("Player wins!")
    elif dealer_total > player_total:
        print("Dealer wins.")
    else:
        print("It's a push (tie)!")

if __name__ == "__main__":
    play_blackjack()
```

## 2.3 Task 3: Implement Multiplayer Support

```python
import random
from enum import Enum

class Suit(Enum):
    HEARTS = "Hearts"
    DIAMONDS = "Diamonds"
    CLUBS = "Clubs"
    SPADES = "Spades"

class Card:
    def __init__(self, suit: Suit, rank: str):
        self.suit = suit
        self.rank = rank
    def __repr__(self):
        return f"Card({self.suit.value}, {self.rank})"

class Deck:
    RANKS = ["Ace", "2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King"]
    def __init__(self):
        self.cards = [Card(suit, rank) for suit in Suit for rank in self.RANKS]
        self.shuffle()
    def shuffle(self):
        random.shuffle(self.cards)
    def draw(self):
        return self.cards.pop() if self.cards else None

class Hand:
    def __init__(self):
        self.cards = []
    def add_card(self, card):
        self.cards.append(card)
    def get_value(self):
        value = 0
        aces = 0
        for card in self.cards:
            if card.rank in ["Jack", "Queen", "King"]: value += 10
            elif card.rank == "Ace":
                aces += 1
                value += 11
            else: value += int(card.rank)
        while value > 21 and aces > 0:
            value -= 10
            aces -= 1
        return value

class Player(Hand):
    def __init__(self, name):
        super().__init__()
        self.name = name
        self.busted = False

def play_multiplayer_blackjack(num_players):
    deck = Deck()
    players = [Player(f"Player {i+1}") for i in range(num_players)]
    dealer = Player("Dealer")

    # Deal initial cards
    for _ in range(2):
        for player in players:
            player.add_card(deck.draw())
        dealer.add_card(deck.draw())

    # Turn-based system for players
    for player in players:
        print(f"\n--- {player.name}'s Turn ---")
        print(f"Hand: {player.cards} (Total: {player.get_value()})")
        
        while player.get_value() < 21:
            action = input(f"{player.name}, do you want to (h)it or (s)tand? ").lower()
            if action == 'h':
                player.add_card(deck.draw())
                print(f"Updated Hand: {player.cards} (Total: {player.get_value()})")
            elif action == 's':
                break
            else:
                print("Invalid input.")
                
        if player.get_value() > 21:
            print(f"{player.name} busts and is out of the game!")
            player.busted = True

    # Check if all players busted
    active_players = [p for p in players if not p.busted]
    if not active_players:
        print("\nAll players busted. Dealer wins by default.")
        return

    # Dealer's turn (plays once all players have stood or busted)
    print(f"\n--- Dealer's Turn ---")
    print(f"Dealer's Hand starts as: {dealer.cards} (Total: {dealer.get_value()})")
    while dealer.get_value() <= 16: # Hit on 16 or less
        print("Dealer hits...")
        dealer.add_card(deck.draw())
        print(f"Dealer's Hand: {dealer.cards} (Total: {dealer.get_value()})")

    # Final Comparisons
    dealer_total = dealer.get_value()
    print("\n=== FINAL RESULTS ===")
    
    if dealer_total > 21:
        print("Dealer busted! All remaining players win automatically.")
        for player in active_players:
            print(f"{player.name} wins!")
    else:
        # Compare dealer with each standing player
        for player in active_players:
            player_total = player.get_value()
            if player_total > dealer_total:
                print(f"{player.name} wins! ({player_total} vs {dealer_total})")
            elif dealer_total > player_total:
                print(f"{player.name} loses to Dealer. ({player_total} vs {dealer_total})")
            else:
                print(f"{player.name} pushes (ties) with Dealer. ({player_total} vs {dealer_total})")


if __name__ == "__main__":
    print("Welcome to Multiplayer Blackjack!")
    try:
        num = int(input("Enter number of players (e.g., 2 or 3): "))
        if num > 0:
            play_multiplayer_blackjack(num)
        else:
            print("Need at least 1 player.")
    except ValueError:
        print("Please enter a valid number.")
```
