## 1. Write a recursive function to compute the sum of the first n natural numbers. Then, analyze its time complexity using Big O notation.
## Tasks:
### (a) Implement a recursive function recursive sum(n) that computes the sum of the first n natural numbers using recursion.
### (b) Determine the base case and recursive step.
### (c) Write a function iterative sum(n) that computes the sum using a loop.
### (d) Compare the time complexity of both functions using Big O notation.
### (e) Explain why recursion is less efficient in this case compared to iteration.

```python
# (a) Python code for recursive sum(n) and iterative sum(n).

def recursive_sum(n):
    if n <= 0:
        return 0
    if n == 1:
        return 1
    return n + recursive_sum(n - 1)

def iterative_sum(n):
    if n <= 0:
        return 0
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

n = int(input("ENTER VALUE OF N"))
print(f"Recursive sum of {n}: {recursive_sum(n)}")
print(f"Iterative sum of {n}: {iterative_sum(n)}")
```

```python
# (b) A written explanation of the time complexity of both functions.

''' 
Recursive Function (recursive_sum): Time Complexity: O(n). 
The function calls itself n times. 
Each call performs a constant amount of work O(1)—specifically, checking the if-statements and performing a single addition. 
Therefore, the total time is proportional to n.
Space Complexity: O(n). 
Every time a recursive call is made, a new frame is added to the system's call stack to store the current state. 
Because it makes n nested calls before reaching the base case, it requires memory proportional to n.

Iterative Function (iterative_sum):
Time Complexity: O(n). 
The for loop runs exactly n times, and the addition operation inside the loop takes constant time O(1).
Space Complexity: O(1). 
The iterative approach only requires a fixed amount of memory to store the total and i variables, regardless of how large n grows.
'''
```

```python
# (c) A comparison between recursion and iteration in this problem.

'''
Comparison Between Recursion and Iteration

1. Recursion uses function calls, whereas iteration uses loops.
2. Both recursive_sum(n) and iterative_sum(n) have O(n) time complexity.
3. Recursion requires O(n) extra space because each recursive call is stored in the call stack.
4. Iteration requires only O(1) space since it uses a few variables.
5. Recursive solutions are generally slower due to function call overhead.
6. Iterative solutions are faster and more memory efficient.
7. Recursion may cause stack overflow for very large values of n.
8. Iteration is preferred for this problem because it is simpler, efficient, and uses less memory.
'''
```

## 2. The Josephus Problem is a famous theoretical problem related to recursion and combinatorics. It is defined as follows:
### • There are n people standing in a circle, numbered from 0 to n− 1.
### • Every k-th person is eliminated in a circular manner until only one person remains.
### • Your task is to implement a recursive function to find the position of the last remaining person (zero-based index).
## Tasks:
### (a) Implement a recursive function josephus(n, remaining.
### (b) Identify the base case and the recursive formula.
### (c) Analyze the time complexity of your recursive implementation using Big O notation.
### (d) Implement an iterative version and compare its efficiency with the recursive version.
### (e) Test the function with different values of n and k.

```python
# Python code for josephus(n, k) using recursion.

def josephus_recursive(n, k):
    if n == 1:
        return 0
    return (josephus_recursive(n - 1, k) + k) % n

def josephus_iterative(n, k):
    result = 0
    for i in range(2, n + 1):
        result = (result + k) % i
    return result

test_cases = [(5, 2), (7, 3), (14, 2)]

print(f"{'n':<5} {'k':<5} | {'Recursive':<10} | {'Iterative':<10}")
print("-" * 40)
for n_val, k_val in test_cases:
    rec = josephus_recursive(n_val, k_val)
    ite = josephus_iterative(n_val, k_val)
    print(f"{n_val:<5} {k_val:<5} | {rec:<10} | {ite:<10}")
```

```python
# Explanation of the recursive logic and base case.

'''
The logic relies on re-indexing the circle after each elimination.

Base Case: If there is only one person (n=1), their position is 0.
Recursive Formula : J(n, k) = (J(n - 1, k) + k) % n 

Explanation: When the first person (at index k-1) is removed, the remaining n-1 people form a new circle. 
To find the winner's original position, we solve the problem for n-1 people and then "shift" that result 
by k to account for the elimination, using the modulo operator (p mod n) to keep the index within the circular bounds.
'''
```

```python
# Big O complexity analysis.

'''
Time Complexity: O(n). 
The function calls itself once for every person until n reaches 1. Each call performs a constant amount of work.

Space Complexity: O(n). 
Because this is not naturally tail-recursive in Python, the system must maintain n frames on the call stack.
'''
```

```python
# Comparison of recursion vs. iteration for solving the problem.

'''
Comparison Between Recursive and Iterative Josephus Solution

1. The recursive approach solves the problem by calling the function repeatedly until only one person remains.
2. The iterative approach solves the problem using a loop without making recursive function calls.
3. Both recursive and iterative methods have O(n) time complexity.
4. The recursive solution requires O(n) space because recursive calls are stored in the call stack.
5. The iterative solution requires O(1) space since it only uses variables.
6. Recursive implementation is easier to understand because it directly follows the mathematical recurrence relation.
7. Iterative implementation is more memory efficient and avoids stack overflow.
8. For large values of n, the iterative approach is generally preferred because it is safer and more efficient.
'''
```

## 3. You need to implement an efficient recursive function to compute ab (where a is the base and b is the exponent) using Exponentiation by Squaring, which reduces the number of recursive calls significantly.
## Tasks:
### (a) Implement a recursive function fast power(a, b) that computes ab using divide and conquer.
### (b) Identify the base case and the recursive formula:
####      • If b is even, use the identity:
####             a^b = (a^(b/2))^2
####      • If b is odd, use the identity:
####             a^b = a × a^(b−1)
### (c) Compare the time complexity of the standard recursive method (O(b)) and the optimized exponentiation by squaring method (O(log b)).
### (d) Implement an iterative version of exponentiation by squaring and compare it with the recursive implementation.
### (e) Test your function with large values of b (e.g., a = 2, b = 1000000) and observe the performance.

```python
# Python code for fast power(a, b) using recursion.

import sys
def fast_power_recursive(a, b):
    if b == 0:
        return 1
    if b < 0:
        return 1 / fast_power_recursive(a, -b)
    if b % 2 == 0:
        half = fast_power_recursive(a, b // 2)
        return half * half
    else:
        return a * fast_power_recursive(a, b - 1)

def fast_power_iterative(a, b):
    if b < 0:
        a = 1 / a
        b = -b
    result = 1
    while b > 0:
        if b % 2 == 1:
            result *= a
        a *= a
        b //= 2
    return result

a, b = 2, 10
print(f"Recursive {a}^{b}: {fast_power_recursive(a, b)}")
print(f"Iterative {a}^{b}: {fast_power_iterative(a, b)}")
```

```python
# Explanation of divide-and-conquer logic used in exponentiation by squaring.

''' 
The Divide and Conquer logic works by breaking the problem into halves rather than subtracting 1 at each step.
Base Case: a^0 = 1. Any number raised to the power of 0 is 1.
Recursive Formulae: If b is even: a^b = (a^{b/2})^2. Instead of b multiplications, we do the work for b/2 and square the result. 
If b is odd: a^b = a \times a^{b-1}. We convert it to an even case so the next step can be halved.
'''
```

```python
# Big O complexity analysis for both naive recursion and optimized recursion.

'''
Big O Complexity Analysis

Naive Recursive Method:

In the normal recursive approach, the function multiplies 'a'
exactly 'b' times.

Example:
power(a, b) = a * power(a, b-1)

Time Complexity:
O(b)

because there are b recursive calls.

Space Complexity:
O(b)

due to the recursive call stack.


Optimized Exponentiation by Squaring:

This method divides the exponent by 2 in each recursive call.

If b is even:
a^b = (a^(b/2))^2

If b is odd:
a^b = a * a^(b-1)

Time Complexity:
O(log b)

because the exponent is reduced by half in every step.

Space Complexity:
O(log b)

due to recursive calls decreasing logarithmically.


Comparison:

1. Naive recursion performs b recursive calls.
2. Optimized recursion performs only log(b) recursive calls.
3. Exponentiation by squaring is much faster for very large exponents.
4. Optimized recursion is more efficient in both time and memory usage.
'''
```

```python
# Comparison with an iterative version of the algorithm.

'''
Comparison Between Recursive and Iterative Exponentiation by Squaring

1. The recursive version solves the problem using recursive function calls, whereas the iterative version uses loops.
2. Both recursive and iterative exponentiation by squaring algorithms have O(log b) time complexity.
3. The recursive approach requires O(log b) space because recursive calls are stored in the call stack.
4. The iterative approach requires O(1) space since it only uses variables and loops.
5. Recursive implementation is shorter and easier to understand.
6. Iterative implementation is more memory efficient and avoids stack overflow.
7. For very large values of b, the iterative method is generally preferred because it is faster and safer in practice.
8. Both methods are significantly more efficient than the naive recursive approach with O(b) time complexity.
'''
```

## Phantom tracking
### A notorious hacker known as The Phantom has been manipulating digital transactions in Bhubaneswar to move stolen money through multiple accounts. The Cyber Crime Unit has intercepted a list of transactions (in thousands of rupees) from different accounts. However, the transactions are unordered, some are duplicates, and The Phantom has hidden key details among normal transactions.
### Your task is to analyze the data step by step, finally leading to the criminal.
#### Objective 1) Identifying Suspicious Transactions
#### The police believe that certain high-value transactions (above a given threshold) are linked to the criminal network.
#### Task: Use Linear Search to extract all suspicious transactions above a certain amount.
#### Given Data:
#### transactions = [120, 45, 300, 220, 90, 600, 130, 75, 800, 500, 350, 40]
#### threshold = 250
#### Your goal is to identify transactions greater than |250,000 to narrow the suspect list.
#### Once you extract the high-value transactions, you will use them for the next step.
#### Objective 2) Organizing the Transaction Data
#### The extracted suspicious transactions from Objective 1 need to be sorted in ascending order for further analysis.
#### Task:
#### Use Selection Sort to sort the suspicious transactions and return a sorted list of transactions.
#### Once sorted, the police will use these transactions to identify frequent recipients.
#### Objective 3) Finding a Specific Transaction
#### A forensic accountant identified a key transaction—one of these amounts was transferred to an account linked to The Phantom.
#### Task:
#### Use Binary Search to check if a specific suspicious transaction exists in the sorted list search amount = 500
#### Return True if the |500,000 transaction is in the suspicious list, otherwise return False.
#### If this transaction exists, it will lead to a secret offshore account linked to The Phantom.
#### Objective 4) Reconstructing the Criminal’s Full Transaction History
#### The police have recovered all transaction records from The Phantom’s secret bank. However, they are completely unorganized and need to be sorted for further analysis.
#### Task:
#### Use Merge Sort to sort all transactions.
#### Input: transactions = [120, 45, 300, 220, 90, 600, 130, 75, 800, 500, 350, 40]
#### Your goal is to return a fully sorted list so the police can analyze the transaction history and track The Phantom.
#### Once sorted, the police notice a repeating pattern of money being sent to a specific name— The Phantom’s real identity!

```python

transactions = [120, 45, 300, 220, 90, 600, 130, 75, 800, 500, 350, 40]
threshold = 250

#Objective 1: Identifying Suspicious Transactions (Linear Search)
def get_suspicious_transactions(data, limit):
    suspicious = []
    for tx in data:
        if tx > limit:
            suspicious.append(tx)
    return suspicious

suspicious_list = get_suspicious_transactions(transactions, threshold)
print(f"Objective 1 - Suspicious Transactions: {suspicious_list}")


#Objective 2: Organizing the Transaction Data (Selection Sort)
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

sorted_suspicious = selection_sort(suspicious_list.copy())
print(f"Objective 2 - Sorted Suspicious List: {sorted_suspicious}")


#Objective 3: Finding a Specific Transaction (Binary Search)
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return True
        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return False

search_amount = 500
found = binary_search(sorted_suspicious, search_amount)
print(f"Objective 3 - Transaction {search_amount} found: {found}")


#Objective 4: Reconstructing Full History (Merge Sort)
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

full_history = merge_sort(transactions)
print(f"Objective 4 - Full Sorted Transaction History: {full_history}")
```
